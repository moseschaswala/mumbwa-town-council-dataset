# CDF Data Integration

This notebook combines processed CSV files only when they contain the same type of CDF data.

## 1. Import Libraries

```python
from pathlib import Path
import sys

import pandas as pd

project_root = Path.cwd()
if project_root.name == "notebooks":
    project_root = project_root.parent

sys.path.append(str(project_root))

from src.paths import PROCESSED_DIR, FINAL_DIR
```

## 2. Choose Files With The Same Data Type

```python
# Example: only choose grant files here.
grant_files = sorted(PROCESSED_DIR.glob("*grant*_clean*.csv"))

print("Grant files found:", len(grant_files))
for file in grant_files:
    print(file.name)
```

## 3. Combine The Files

```python
if grant_files:
    grant_datasets = [pd.read_csv(file, sep="|") for file in grant_files]
    grants = pd.concat(grant_datasets, ignore_index=True)
    grants.head()
else:
    grants = pd.DataFrame()
    print("No grant files found yet.")
```

## 4. Save Final Dataset

```python
if not grants.empty:
    output_file = FINAL_DIR / "db-unza26-csc4792-mumbwa_town_council_cdf_grants.csv"
    grants.to_csv(output_file, index=False, sep="|")
    print("Saved to:", output_file)
else:
    print("Nothing to save yet.")
```
