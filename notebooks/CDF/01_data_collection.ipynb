# CDF Data Collection

This notebook starts from the Mumbwa Town Council website and collects the CDF PDF documents used in this project.

## 1. Import Libraries

```python
from pathlib import Path
import sys

import pandas as pd

project_root = Path.cwd()
if project_root.name == "notebooks":
    project_root = project_root.parent

sys.path.append(str(project_root))

from src.collector import CDF_TRACKER_URL, find_pdf_links, save_source_list, download_pdfs
from src.paths import RAW_PDF_DIR, RECONSTRUCTED_PDF_DIR, SOURCE_LIST_FILE
```

## 2. Mumbwa Town Council CDF Tracker Page

```python
CDF_TRACKER_URL
```

## 3. Find CDF PDF Links On The Website

```python
# This cell needs an internet connection.
pdf_links = find_pdf_links(CDF_TRACKER_URL)
source_df = pd.DataFrame(pdf_links)

print("CDF PDF links found:", len(source_df))
source_df
```

## 4. Save The Source Link List

```python
source_file = save_source_list(pdf_links, SOURCE_LIST_FILE)
print("Saved source list to:", source_file)
```

## 5. Download Original PDFs

```python
# Existing PDFs are not overwritten.
downloaded_files = download_pdfs(pdf_links, RAW_PDF_DIR)

print("PDF files available:", len(downloaded_files))
for pdf_file in downloaded_files:
    print(pdf_file.name)
```

## 6. Check Original PDFs Saved Locally

```python
pdf_files = sorted(RAW_PDF_DIR.glob("*.pdf"))

print("Original PDFs found:", len(pdf_files))
for pdf_file in pdf_files:
    print(pdf_file.name)
```

## 7. Check Recreated PDFs

```python
recreated_files = sorted(RECONSTRUCTED_PDF_DIR.glob("*.pdf"))

print("Recreated PDFs found:", len(recreated_files))
for pdf_file in recreated_files:
    print(pdf_file.name)
```
