# CDF Data Validation

This notebook checks cleaned CDF data and shows warnings for records that need manual checking.

## 1. Import Libraries

```python
from pathlib import Path
import sys

import pandas as pd

project_root = Path.cwd()
if project_root.name == "notebooks":
    project_root = project_root.parent

sys.path.append(str(project_root))

from src.paths import PROCESSED_DIR
from src.validator import check_missing_values, check_duplicates, check_required_columns, check_amount_column
```

## 2. Load Cleaned Dataset

```python
processed_files = sorted(PROCESSED_DIR.glob("*.csv"))

print("Processed CSV files found:", len(processed_files))
for file in processed_files:
    print(file.name)
```

```python
input_file = processed_files[0] if processed_files else None

if input_file is None:
    print("No processed CSV files found.")
    df = pd.DataFrame()
else:
    df = pd.read_csv(input_file, sep="|")
    print("Loaded:", input_file.name)

df.head()
```

## 3. Run Validation Checks

```python
if not df.empty:
    check_missing_values(df)
    check_duplicates(df)
else:
    print("No data loaded yet.")
```

## 4. Check Important Columns

```python
# Change this list to match the dataset you are checking.
required_columns = ["constituency", "year", "source_document"]

if not df.empty:
    check_required_columns(df, required_columns)
else:
    print("No data loaded yet.")
```

## 5. Check Amount Column

```python
# Change the column name if your amount column has a different name.
if not df.empty:
    check_amount_column(df, "amount")
else:
    print("No data loaded yet.")
```
