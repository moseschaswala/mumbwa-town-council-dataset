```python
import pandas as pd
import re
import pdfplumber

prog_rows = []

budget_2026_file.seek(0)

with pdfplumber.open(budget_2026_file) as pdf:
    # --- 1. Parse Table 2 (Page 7): Programme Allocations ---
    pg7_text = pdf.pages[6].extract_text() or ""
    for line in pg7_text.split("\n"):
        line = line.strip()
        # Look for lines starting with 1-2 digits, text, and 3 financial figures
        match = re.match(r"^(\d{1,2})\s+(.*?)\s+([\d,]+|-)\s+([\d,]+|-)\s+([\d,]+|-)$", line)
        if match and "Head Total" not in line and "Code" not in line:
            p_code = match.group(1)
            p_name = match.group(2).strip()
            amt_2026 = match.group(5).replace(",", "").replace("-", "0")

            prog_rows.append({
                "year": 2026,
                "financial_type": "Programme Budget",
                "category/code": f"Prog {p_code}",
                "description": p_name,
                "amount": float(amt_2026),
                "source": "2026 Approved Budget PDF (Page 7)"
            })

    # --- 2. Parse Table 3 (Pages 8-9): Sub-Programme Allocations ---
    sub_text = (pdf.pages[7].extract_text() or "") + "\n" + (pdf.pages[8].extract_text() or "")
    for line in sub_text.split("\n"):
        # Clean cid artifacts
        line = re.sub(r"\(cid:\d+\)", "", line).strip()

        # Matches lines starting with 3-digit sub-programme code (e.g. 779, 003, 038)
        # and captures the final number (2026 estimate) on the line
        match = re.match(r"^(\d{3})\s+(.*?)\s+([\d,\s\-\(\)]+)$", line)
        if match:
            sp_code = match.group(1)
            rest = match.group(2).strip()

            # Extract all numerical chunks from the end of the line
            numbers = re.findall(r"[\d,]+|\(0\)", line)
            if numbers:
                amt_2026_raw = numbers[-1].replace(",", "").replace("(0)", "0")
                # Remove trailing numbers from description if caught
                sp_name = re.sub(r"\s+[\d,]+.*$", "", rest).strip()

                prog_rows.append({
                    "year": 2026,
                    "financial_type": "Sub-Programme Budget",
                    "category/code": f"SubProg {sp_code}",
                    "description": sp_name,
                    "amount": float(amt_2026_raw),
                    "source": "2026 Approved Budget PDF (Pages 8-9)"
                })

# Append to master budget_2026_df
df_programmes = pd.DataFrame(prog_rows)
budget_2026_df = pd.concat([budget_2026_df, df_programmes], ignore_index=True)

print(f"Successfully extracted {len(df_programmes)} programme/sub-programme rows.")
print(f"Total dataset records: {len(budget_2026_df)}")
display(budget_2026_df.tail(20))
```

```python
import pandas as pd
import re
import pdfplumber

prog_econ_rows = []

budget_2026_file.seek(0)

# Regex patterns
prog_header_pattern = re.compile(r"^Programme\s+(\d{1,2})\s*:\s*(.*)$")
# Matches economic lines like "01Personal Emoluments ... 1,384,057" or "02General Operations ... 11,084,052"
econ_line_pattern = re.compile(r"^(\d{2})\s*([A-Za-z\s\(\)/-]+?)\s+[\d,\s\-\(\)]+$")

current_prog_num = None
current_prog_title = None

with pdfplumber.open(budget_2026_file) as pdf:
    # Scan through pages 10 to 52 for granular line-item breakdowns
    for page_idx in range(9, len(pdf.pages)):
        text = pdf.pages[page_idx].extract_text() or ""
        lines = text.split("\n")

        for line in lines:
            line = line.strip()

            # 1. Detect Programme Header
            prog_match = prog_header_pattern.match(line)
            if prog_match:
                current_prog_num = prog_match.group(1)
                current_prog_title = prog_match.group(2).strip()
                continue

            # 2. Detect Economic Classification lines under Table 4
            econ_match = econ_line_pattern.match(line)
            if econ_match and current_prog_title:
                code_prefix = econ_match.group(1)
                desc = econ_match.group(2).strip()

                # Exclude summary/total lines or headers
                if desc.lower() in ["programme total", "economic classification", "approved", "estimates"]:
                    continue

                # Extract all numbers from the line; the last number is 2026 Estimate
                numbers = re.findall(r"[\d,]+|\(0\)", line)
                if numbers:
                    amt_str = numbers[-1].replace(",", "").replace("(0)", "0")
                    try:
                        amt = float(amt_str)
                        prog_econ_rows.append({
                            "year": 2026,
                            "financial_type": f"Programme {current_prog_num} Expenditure",
                            "category/code": f"Prog {current_prog_num} ({code_prefix})",
                            "description": f"{current_prog_title} - {desc}",
                            "amount": amt,
                            "source": f"2026 Approved Budget PDF (Page {page_idx + 1})"
                        })
                    except ValueError:
                        continue

# Append to master dataset
df_prog_econ = pd.DataFrame(prog_econ_rows)
budget_2026_df = pd.concat([budget_2026_df, df_prog_econ], ignore_index=True)

print(f"Extracted {len(df_prog_econ)} detailed programme economic classification items.")
print(f"Total dataset records: {len(budget_2026_df)}")
display(budget_2026_df.tail(25))
```

```python
import pandas as pd
import re
import pdfplumber

budget_2025_df = pd.DataFrame()
prog_rows_2025 = []

budget_2025_file.seek(0)

with pdfplumber.open(budget_2025_file) as pdf:
    # --- 1. Parse Revenue Tables (Pages 3 to 6) ---
    current_major_head = ""

    for page_idx in range(2, 6):  # Pages 3 to 6
        text = pdf.pages[page_idx].extract_text() or ""
        lines = text.split("\n")

        for line in lines:
            line = re.sub(r"\(cid:\d+\)", "", line).strip()

            # Detect major category headers (e.g., "01 Local taxes/rates", "02 Fees and Charges")
            cat_match = re.match(r"^(\d{2})\s+([A-Za-z\s/]+)$", line)
            if cat_match:
                current_major_head = cat_match.group(2).strip()
                continue

            # Match revenue sub-items starting with 3-digit code
            rev_match = re.match(r"^(\d{3})\s+(.*?)\s+([\d,]+|-)\s+([\d,]+|-)\s+([\d,]+|-)$", line)
            if rev_match:
                code = rev_match.group(1)
                desc = rev_match.group(2).strip()
                amt_2025 = rev_match.group(3).replace(",", "").replace("-", "0")

                prog_rows_2025.append({
                    "year": 2025,
                    "financial_type": "Revenue Budget",
                    "category/code": f"{code} ({current_major_head})",
                    "description": desc,
                    "amount": float(amt_2025),
                    "source": f"2025 Approved Budget PDF (Page {page_idx + 1})"
                })

    # --- 2. Parse Economic Classification (Page 7) ---
    pg7_text = pdf.pages[6].extract_text() or ""
    for line in pg7_text.split("\n"):
        econ_match = re.match(r"^(\d{2})\s+(.*?)\s+[\d,\s\-\(\)]+\s+([\d,]+|-)$", line.strip())
        if econ_match and "ECONOMIC CLASSIFICATION" not in line:
            code = econ_match.group(1)
            desc = econ_match.group(2).strip()
            amt_2025 = econ_match.group(3).replace(",", "").replace("-", "0")

            prog_rows_2025.append({
                "year": 2025,
                "financial_type": "Expenditure Budget (Economic Classification)",
                "category/code": code,
                "description": desc,
                "amount": float(amt_2025),
                "source": "2025 Approved Budget PDF (Page 7)"
            })

    # --- 3. Parse Table 2: Programme Allocations (Page 8) ---
    pg8_text = pdf.pages[7].extract_text() or ""
    for line in pg8_text.split("\n"):
        prog_match = re.match(r"^(\d{1,2})\s+(.*?)\s+[\d,\s\-\(\)]+\s+([\d,]+|-)$", line.strip())
        if prog_match and "Programme" not in line and "Head Total" not in line:
            p_code = prog_match.group(1)
            p_name = prog_match.group(2).strip()
            amt_2025 = prog_match.group(3).replace(",", "").replace("-", "0")

            prog_rows_2025.append({
                "year": 2025,
                "financial_type": "Programme Budget",
                "category/code": f"Prog {p_code}",
                "description": p_name,
                "amount": float(amt_2025),
                "source": "2025 Approved Budget PDF (Page 8)"
            })

    # --- 4. Parse Table 3: Sub-Programme Allocations (Pages 9-11) ---
    sub_text = ""
    for p in range(8, min(11, len(pdf.pages))):
        sub_text += (pdf.pages[p].extract_text() or "") + "\n"

    for line in sub_text.split("\n"):
        line = re.sub(r"\(cid:\d+\)", "", line).strip()
        sp_match = re.match(r"^(\d{3})\s+(.*?)\s+[\d,\s\-\(\)]+\s+([\d,]+|\(0\)|-)$", line)
        if sp_match:
            sp_code = sp_match.group(1)
            rest = sp_match.group(2).strip()

            numbers = re.findall(r"[\d,]+|\(0\)", line)
            if numbers:
                amt_2025_raw = numbers[-1].replace(",", "").replace("(0)", "0")
                sp_name = re.sub(r"\s+[\d,]+.*$", "", rest).strip()

                prog_rows_2025.append({
                    "year": 2025,
                    "financial_type": "Sub-Programme Budget",
                    "category/code": f"SubProg {sp_code}",
                    "description": sp_name,
                    "amount": float(amt_2025_raw),
                    "source": "2025 Approved Budget PDF (Pages 9-11)"
                })

# Save to budget_2025_df
budget_2025_df = pd.DataFrame(prog_rows_2025)

print(f"Successfully extracted {len(budget_2025_df)} 2025 budget entries.")
display(budget_2025_df.head(15))
display(budget_2025_df.tail(15))
```

```python
import pandas as pd
import re
import pdfplumber

stmt_2024_rows = []

stmt_2024_file.seek(0)

def clean_ocr_num(val_str):
    if not val_str:
        return 0.0
    # Clean OCR character replacements in numeric columns
    cleaned = (
        val_str.replace("O", "0")
        .replace("o", "0")
        .replace("I", "1")
        .replace("l", "1")
        .replace("S", "5")
        .replace("s", "5")
        .replace("Z", "2")
        .replace("z", "2")
        .replace("g", "9")
        .replace("J", "1")
        .replace("a", "0")
        .replace("t", "1")
    )
    cleaned = re.sub(r"[^\d.]", "", cleaned)
    try:
        return float(cleaned) if cleaned else 0.0
    except ValueError:
        return 0.0

with pdfplumber.open(stmt_2024_file) as pdf:
    # --- 1. Page 15: Main Statement of Cash Receipts and Payments ---
    pg15_text = pdf.pages[14].extract_text() or ""
    current_section = "Receipt"

    for line in pg15_text.split("\n"):
        line = line.strip()
        if "PAYMENTS" in line:
            current_section = "Payment"
            continue

        tokens = line.split()
        num_tokens = [t for t in tokens if re.search(r"\d", t)]
        desc_match = re.match(r"^([A-Za-z\s&/-]+)", line)

        if desc_match and num_tokens:
            desc = desc_match.group(1).strip()
            if desc.upper() not in ["TOTAL RECEIPTS", "TOTAL PAYMENTS", "RECEIPTS", "PAYMENTS", "STATEMENT OF CASH RECEIPTS AND PAYMENTS FOR THE YEAR ENDED 3IST"]:
                amt_2024 = clean_ocr_num(num_tokens[0] if len(num_tokens) == 1 else num_tokens[-2])
                if amt_2024 > 0:
                    stmt_2024_rows.append({
                        "year": 2024,
                        "financial_type": f"Actual {current_section} (Main Statement)",
                        "category/code": "Main Statement",
                        "description": desc,
                        "amount": amt_2024,
                        "source": "2024 Financial Statements PDF (Page 15)"
                    })

    # --- 2. Page 17: LGEF Statement ---
    pg17_text = pdf.pages[16].extract_text() or ""
    for line in pg17_text.split("\n"):
        line = line.strip()
        if any(kw in line.lower() for kw in ["salaries", "service", "goods", "administrative", "infrastructure", "rehabilitation", "asset"]):
            parts = line.split()
            desc = " ".join([p for p in parts if not re.search(r"\d", p)])
            nums = [p for p in parts if re.search(r"\d", p)]
            if nums and desc:
                amt = clean_ocr_num(nums[0])
                if amt > 0:
                    stmt_2024_rows.append({
                        "year": 2024,
                        "financial_type": "Actual LGEF Performance",
                        "category/code": "LGEF",
                        "description": f"LGEF - {desc.strip()}",
                        "amount": amt,
                        "source": "2024 Financial Statements PDF (Page 17)"
                    })

    # --- 3. Page 18: CDF Statement ---
    pg18_text = pdf.pages[17].extract_text() or ""
    for line in pg18_text.split("\n"):
        line = line.strip()
        if any(kw in line.lower() for kw in ["infrastructure", "rehabilitation", "asset", "bursaries", "empowerment", "administrative", "disaster"]):
            parts = line.split()
            desc = " ".join([p for p in parts if not re.search(r"^\d+\([a-z]\)$", p) and not re.search(r"^\d+$", p)])
            nums = [p for p in parts if re.search(r"\d{3,}", p)]
            if nums and desc:
                amt = clean_ocr_num(nums[0])
                if amt > 0:
                    stmt_2024_rows.append({
                        "year": 2024,
                        "financial_type": "Actual CDF Performance",
                        "category/code": "CDF",
                        "description": f"CDF - {desc.strip()}",
                        "amount": amt,
                        "source": "2024 Financial Statements PDF (Page 18)"
                    })

    # --- 4. Page 20: Sector Grants ---
    pg20_text = pdf.pages[19].extract_text() or ""
    for line in pg20_text.split("\n"):
        line = line.strip()
        if any(kw in line.lower() for kw in ["goods", "non-financial", "infrastructure", "rehabilitation"]):
            parts = line.split()
            desc = " ".join([p for p in parts if not re.search(r"\d", p)])
            nums = [p for p in parts if re.search(r"\d", p)]
            if nums and desc:
                amt = clean_ocr_num(nums[0])
                if amt > 0:
                    stmt_2024_rows.append({
                        "year": 2024,
                        "financial_type": "Actual Sector Grant Performance",
                        "category/code": "Sector Grant",
                        "description": f"Sector Grant - {desc.strip()}",
                        "amount": amt,
                        "source": "2024 Financial Statements PDF (Page 20)"
                    })

# Export DataFrame to CSV
df_stmt_2024 = pd.DataFrame(stmt_2024_rows)
output_2024_filename = "db-unza26-csc4792-mumbwa_2024_actual_financial_statements.csv"
df_stmt_2024.to_csv(output_2024_filename, sep="|", index=False)

print(f"Successfully extracted {len(df_stmt_2024)} actual 2024 financial records.")
print(f"Saved to '{output_2024_filename}'.")
display(df_stmt_2024)
```

```python
import os
import re
import requests
import urllib3
from bs4 import BeautifulSoup
import pandas as pd

urllib3.disable_warnings(urllib3.exceptions.InsecureRequestWarning)

# Target Council Document URLs / Categories
base_url = "https://www.mumbwacouncil.gov.zm"
target_pages = [
    f"{base_url}/documents/",
    f"{base_url}/category/financial-reports/",
    f"{base_url}/category/budgets/",
    f"{base_url}/constituency-development-fund-cdf/"
]

headers = {'User-Agent': 'Mozilla/5.0'}

discovered_files = []

# Step 1: Discover all financial PDF links across the website
print("=== DISCOVERING FINANCIAL DOCUMENTS ON MUMBWA COUNCIL WEBSITE ===")

for page_url in target_pages:
    try:
        res = requests.get(page_url, headers=headers, verify=False, timeout=10)
        if res.status_code == 200:
            soup = BeautifulSoup(res.text, 'html.parser')
            for a in soup.find_all('a', href=True):
                href = a['href']
                text = a.get_text(strip=True)
                if href.endswith('.pdf') or 'budget' in href.lower() or 'financial' in href.lower():
                    discovered_files.append({
                        "title": text or os.path.basename(href),
                        "url": href if href.startswith('http') else base_url + href
                    })
    except Exception as e:
        print(f"Skipping {page_url}: {e}")

# Deduplicate discovered links
df_links = pd.DataFrame(discovered_files).drop_duplicates(subset=['url'])
print(f"\nFound {len(df_links)} total financial/publication PDFs on the website:\n")
display(df_links)

# Save link inventory
df_links.to_csv("mumbwa_council_all_financial_pdf_index.csv", sep="|", index=False)
```
