```python
import nbformat
from pathlib import Path

notebook_files = [
    "00_mumbwa_town_council_start.ipynb",
    "01_data_collection.ipynb",
    "02_pdf_extraction.ipynb",
    "03_data_cleaning.ipynb",
    "04_data_integration.ipynb",
    "05_data_validation.ipynb",
]

merged = nbformat.v4.new_notebook()
merged.cells = []

for filename in notebook_files:
    print(f"Merging: {filename}")

    with open(filename, "r", encoding="utf-8") as f:
        notebook = nbformat.read(f, as_version=4)

    merged.cells.extend(notebook.cells)

output_file = "CSC4792_Mumbwa_Town_Council_Final.ipynb"

with open(output_file, "w", encoding="utf-8") as f:
    nbformat.write(merged, f)

print(f"\nDone! Created: {output_file}")
```
