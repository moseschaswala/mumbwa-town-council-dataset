# CDF Data Cleaning

This notebook cleans one extracted CDF CSV file at a time.

## 1. Import Libraries

```python
from pathlib import Path
import sys

import pandas as pd

project_root = Path.cwd()
if project_root.name == "notebooks":
    project_root = project_root.parent

sys.path.append(str(project_root))

from src.cleaner import clean_grants, clean_community_projects, clean_skills_bursaries, clean_secondary_bursaries
from src.cleaner import clean_all_extracted_csv_files
from src.paths import EXTRACTED_DIR, PROCESSED_DIR
```

## 2. Load Dataset

```python
csv_files = sorted(EXTRACTED_DIR.glob("*.csv"))

print("CSV files found:", len(csv_files))
for csv_file in csv_files:
    print(csv_file.name)
```

```python
input_file = csv_files[0] if csv_files else None

if input_file is None:
    print("No CSV files found in data/extracted.")
    df = pd.DataFrame()
else:
    df = pd.read_csv(input_file, sep="|")
    print("Loaded:", input_file.name)
```

## 4. Clean Dataset

```python
# Change this function if the file is not a grants dataset.
df_clean = clean_grants(df) if not df.empty else df
df_clean.head()
```

## 5. Check The Result

```python
print("Rows before:", len(df))
print("Rows after:", len(df_clean))
df_clean.isna().sum()
```

## 6. Save Clean Dataset

```python
if input_file is not None and not df_clean.empty:
    output_file = PROCESSED_DIR / f"{input_file.stem}_clean.csv"
    number = 2
    while output_file.exists():
        output_file = PROCESSED_DIR / f"{input_file.stem}_clean_{number}.csv"
        number += 1

    df_clean.to_csv(output_file, index=False, sep="|")
    print("Saved to:", output_file)
else:
    print("Nothing to save yet.")
```
